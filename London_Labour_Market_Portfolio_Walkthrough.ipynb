```python
import pandas as pd, numpy as np
panel = pd.read_csv('phase07_labour_market_analysis/A05_Borough_Diagnostic_Panel.csv')
panel.head()
```

## Priority methodology
Higher unemployment and inactivity increase priority; lower jobs density increases priority. Indicators are z-standardised and combined using equal baseline weights, with sensitivity scenarios.


```python
metrics=['unemployment_rate_pct','inactivity_rate_pct','jobs_density']
df=panel[['borough_code','borough_name']+metrics].dropna().copy()
X=pd.DataFrame({m: ((-1 if m=='jobs_density' else 1)*(df[m]-df[m].mean())/df[m].std(ddof=0)) for m in metrics})
df['priority_score']=X.mean(axis=1)
df['priority_rank']=df.priority_score.rank(ascending=False,method='min').astype(int)
df.sort_values('priority_rank').head(10)
```

## Interpretation
The composite score is a relative analytical priority measure, not a causal estimate or funding entitlement. Cluster membership is used as contextual segmentation rather than as a component of the score.
